## Simple Matching

Write a program that asks the user to enter "red", "green", or "blue".
Use match to print a message like "You chose red!", "You chose green!", or "You chose blue!".
If the input doesnâ€™t match, print "Unknown color".


```python
color = input('Enter a color: ')
match color:
    case 'red':
        print(f'You chose color {color}!')
    case 'green':
        print(f'You chose color {color}!')
    case 'blue':
        print(f'You chose color {color}!')
    case _:
        print(f'You chose your own color.')
```

## Matching Multiple Values

Ask the user to enter a day of the week.
Use match to print "Weekend" if the day is Saturday or Sunday, otherwise print "Weekday".
(Hint: lowercase the input first with .lower().)

```python
day = input('Enter a day of the week: ')

match day.lower():
    case 'saturday' | 'sunday':
        print('Weekend')
    case 'monday' | 'tuesday' | 'wednesday' | 'thursday' | 'friday':
        print('Weekday')
    case _:
        print('Invalid day.')



        
```

## Variable Binding

Given a tuple (x, y), use match to check:

(0, 0) â†’ "Origin"

(x, 0) â†’ "On the X-axis"

(0, y) â†’ "On the Y-axis"

(x, y) â†’ "Point at x=..., y=..."

Try it with different values.


```python
point = (2, 3)

match point:
    case (0, 0):
        print('Origin')
    case (x, 0):  # Any point on X-axis
        print(f'On the X-axis at x={x}')
    case (0, y):  # Any point on Y-axis
        print(f'On the Y-axis at y={y}')
    case (x, y):  # Any other point
        print(f'Point at x={x}, y={y}')
```

## Guards (with if)

Write a program that asks the user for an integer.
Use match to check:

If the number is negative

If the number is even

Otherwise, treat it as odd

```python
num = int(input('Enter a number: '))
match num:
    case n if n < 0:
        print(f'{n} is negative.')
    case n if n % 2 == 0:
        print(f'{n} is even.')
    case _:
        print(f'{num} is odd.')
```

## Combining Patterns

Write a program that asks the user to enter a number between 1 and 10.
Use match to:

Print "One" if the number is 1

Print "Two or Three" if the number is 2 or 3

Print "Greater than 5" if the number is > 5

Print "Other number" otherwise

```python
num = int(input('Enter a number: '))
match num:
    case n if n == 1:
        print(f'It is One.')
    case n if n == 2 or n == 3: #case 2 | 3:
        print('Two' or 'Three.')
    case n if n  > 5:
        print(f'It is Greater than five.')
    case _:
        print(f'It is Other number.')
```

## Matching Sequences

Given a list of names, use match to:

Print "Two names: ..." if the list has exactly two items

Print "Three names: ..." if the list has exactly three items

Print "Many names: ..." if the list has more than three items

Test with:

["Alice", "Bob"]
["Alice", "Bob", "Charlie"]
["Alice", "Bob", "Charlie", "David"]



```python
# names = [['Alice', 'Bob'] ['Alice', 'Bob', 'Charlie'] ['Alice', 'Bob', 'Charlie', 'David']]
#Test case 1: Two names

names = ['Alice', 'Bob']
match names:
    case [_, _]:
        print(f'Two numbers: {names}.')
    case [_, _, _]:
        print(f'Three names: {names}.')
    case [_, *_] if len(names) > 3:
        print(f'many names: {names}.')

#Test case 2: Three Names
names = ['Alice', 'Bob', 'Charlie']
match names:
    case [_, _]:
        print(f'Two names: {names}.')
    case [_, _, _]:
        print(f'Three names: {names}')
    case [_, *_] if len(names) > 3:
        print(f'Many names: {names}')

#Test case 3: Four or more names
names = ['Alice', 'Bob', 'Charlie', 'David']
match names:
    case [_, _]:
        print(f'Two names: {names}')
    case [_,_,_]:
        print(f'Three names: {names}')
    case [_, _] if len(names) > 3:
        print(f'many names: {names}.')
```
