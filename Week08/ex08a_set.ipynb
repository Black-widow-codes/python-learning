#### 8
Find the union of a and b

```python
a = {'rob', 'james', 'susan', 'michael', 'rob'} # Create a set from a list of names
b = {'jim', 'michael', 'susanne', 'rob', 'lisa', 'nick'}  # Create another set from a list of names
a.union(b) # Using the set union method to find the combination of sets a and b
a | b # Using the set union operator | to find the combination of sets a and b
print(a.union(b)) # Print the union of sets a and b
print(a | b) # Print the union of sets a and b
```

#### 9
Find the intersection of a and b

```python
a = {'rob', 'james', 'susan', 'michael', 'rob'} # Create a set from a list of names
b = {'jim', 'michael', 'susanne', 'rob', 'lisa', 'nick'}  # Create another set from a list of names
a & b # Intersection or common elements of sets a and b using the set intersection operator &
a.intersection(b) # Using the set intersection method to find the common elements of sets a and b
print(a & b) # Print the intersection of sets a and b
print(f"The intersection of sets a and b is: {a.intersection(b)}") # Print the intersection of sets a and b
```

#### 13
Create two sets of letters from the words "python" and "typhoon".

1.  Find letters they have in common.
1.  Find letters unique to "python".
1.  Find letters unique to "typhoon".
1.  Find all unique letters from both words.

```python
python_letters = set('python')
typhoon_letters = set('typhoon')

common = python_letters & typhoon_letters # Find the common letters between the two strings using set intersection
print(f"The common letters between python and typhoon are: {common}")

unique_to_python = python_letters - typhoon_letters # Find the letters that are in 'python' but not in 'typhoon' using set difference
print(f"The letters that are in 'python' but not in 'typhoon' are: {unique_to_python}")

unique_to_typhoon = typhoon_letters - python_letters # Find the letters that are in 'typhoon' but not in 'python' using set difference
print(f"The letters that are in 'typhoon' but not in 'python' are: {unique_to_typhoon}")

all_unique = python_letters | typhoon_letters # Find all unique letters in both strings using set union
print(f"All unique letters in 'python' and 'typhoon' are: {all_unique}")
```

#### 14
Use set comprehension to create all pairs (x, y) where x is even and y is odd, for x in 0-4 and y in 0-4.

```python
x = {x for x in range(5) if x % 2 == 0}  # even numbers: {0, 2, 4}
y = {y for y in range(5) if y % 2 != 0}  # odd numbers: {1, 3}

pairs = {(x, y) for x in range(5) if x % 2 == 0 for y in range(5) if y % 2 != 0}
print(pairs)
```
