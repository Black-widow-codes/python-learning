#### 6
Given a dictionary of student names and their marks, write code to:

1.  print the highest score
1.  print the name of the student with the lowest score

```python
student_names = {'Alex': 80, 'John': 45, 'Suzan': 90, 'Ivan': 76}
print(student_names)

#Highest score
#max(student_name.value()) gets the max value
highest_score = max(student_names.values())
print(f"Highest score: {highest_score}")

#Name of the student with lowest score
#min(student_names, key=student_names.get()) finds the key whose value is min.
lowest_student = min(student_names, key=student_names.get)
print(f"Student with lowest score: {lowest_student}")



# #print the highest score by looping
# highest = 0
# for score in student_names.values():
#     if score > highest:
#         highest = score
# print(f"Highest score: {highest}")

```

### Dictionary Comprehension

#### 8
Write a dictionary comprehension that maps numbers from 1 to 10 to either 'even' or 'odd'.

```python
a = {num: 'even' if num % 2 == 0 else 'odd' for num in range(1, 11)}
print(a)
```

#### 13
Create a dictionary where the keys are words in a sentence and the values are the number of times each word appears.

```python
from collections import Counter

sentence = 'the old lion and the hyna the old lion and the hyna'
words = sentence.split()
word_count = dict(Counter(words))
print(word_count)


sentence = 'the old lion and the hyna the old lion and the hyna'
words = sentence.split()

#count frequency of each word
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```
