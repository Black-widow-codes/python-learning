#### 1: Basic try-except
Write a program that asks the user for two numbers and divides the first by the second. Handle the case where the user enters zero as the second number.


```python
try:
    user = input('Enter 2 numbers: ')

    #divides the first number by the second.
    numbers = user.split() # Split the input string into a list of strings
    num1 = int(numbers[0]) # Convert the first string to an integer
    num2 = int(numbers[1]) # Convert the second string to an integer
    result = num1 / num2 # Perform the division
    print(f'{num1} divided by {num2} is {result}')

except ZeroDivisionError: # Catch the error if the second number is zero
    print('Error: cannot divide by zero.')

except ValueError: # Catch the error if the input is not a valid integer
    print('Error: please enter a valid number.')

except Exception:
    print('everything else')
```

#### 3: List Index Safety
Write a function safe_get(lst, index) that returns the element at the given index if it exists, or returns None if the index is out of range. Use exception handling to implement this.


```python
def safe_get(lst, index):
    try:
        return lst[index] # Attempt to access the element at the given index
    except IndexError: # Catch the error if index is out of range
        return 'None: Index out of range.'
    
my_list = [10, 20, 30]
print(safe_get(my_list, 0)) 
print(safe_get(my_list, 2))
print(safe_get(my_list, 5))
print(safe_get(my_list, -1))
    
```

#### 4: Dictionary Lookup
Write a program that stores student grades in a dictionary. Ask the user for a student name and display their grade. Handle the case where the student name doesn't exist in the dictionary.


```python
try:
    students_grades = {'bryan': 90, 'mike': 70, 'ronnie': 91}
    student_name = input('Enter a student name:')
    if student_name not in students_grades:
        raise KeyError('Name not found.')

    print(f'Student name: {student_name} has a grade of {students_grades[student_name]}')

except KeyError as e:
    print(f'Error: {e}')

```

#### 5: Multiple Exception Types
Write a program that prompts the user to enter a password. Raise a `ValueError` if the password is less than 8 characters. Raise a `TypeError` if the password contains any spaces. Use try-except blocks to handle these errors and print appropriate messages.


```python
try:
    user_input = input('Enter password. It should be 8 or more characters long: ')

    if len(user_input) < 8:
        raise ValueError('Password must at least be 8 characters long.')
    
    if ' ' in user_input:
        raise TypeError('Password should not contain spaces.')

    print('You successfully logged in.')

except ValueError as e:
    print(f'Error: {e}')

except TypeError as e:
    print(f'Error: {e}')

    


```

#### 7: Age Validator with else and finally
Write a program that asks for a user's age and validates it. Use:

try to convert the input to an integer
except to handle invalid input
else to print a success message
finally to print "Validation complete"

Also check that the age is between 0 and 120.


```python
try:
    user_age = int(input('Enter your age: ')) #Attempts to convert input to int and validate
    
    if user_age not in range(0, 120):
        raise ValueError('Enter age between 0 and 120.')
except ValueError as e: #Runs if valueError is raise
    print(f'Error: {e}')
else:
    print(f'Your age is {user_age}.') #Runs only if no exception occurred in the try block
finally:
    print('Is that your age?') #Always runs

```
