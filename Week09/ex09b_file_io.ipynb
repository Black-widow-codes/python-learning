#### 1: File Modes and Attributes
-   Open a file called `hello.txt` in write mode, write `'Hello, Python!'`, and close it. Then reopen it in read mode and print its contents.
-   Open the same file and print out:
    -   the file name
    -   the mode it was opened in
    -   whether the file is closed or not

```python
# Step 1: Write to file
with open('hello.txt', 'w') as f:  
    f.write('Hello, Python!')

# Step 2: Read from file and print contents
with open('hello.txt', 'r') as f:
    print(f'Contents: {f.read()}')

# Step 3: Open file and print attributes
with open('hello.txt', 'r') as f:
    print(f'File name: {f.name}')
    print(f'Mode: {f.mode}')
    print(f'Closed: {f.closed}')
print(f'After closing - Closed: {f.closed}')
```

#### 2: File Writing
-   Write your 5 most **favorite movie titles** (one per line) into a file called `movies.txt`.

#### 3: File Reading
Write a program that asks the user for a filename and tries to open and read it. If the file doesn't exist, print a friendly error message. Ensure the file object is closed regardless of success or failure.   
Handle exceptions for:  
-   File not found.  
-   Permission errors.   
Respond with clear messages for each error type.



```python
filename = input('Enter a filename to read: ')

try:
    with open(filename, 'r') as f: #Closes file even if error occurs
        content = f.read()
        print(f'File contents:\n{content}')
# except FileNotFoundError:
#     print(f' Error: The file "{filename}" does not exist.')
# except PermissionError:
#     print(f' Error: You do not have permission to read "{filename}".')
# except Exception as e:
#     print(f' Unexpected error: {e}')
except Exception:
    print('Error.')
else:
    print(' File read successfully!')
finally:
    print('Program finished.')
```

#### 4: File Reading
-   Read the contents of movies.txt and print them all at once.
-   Read movies.txt line by line and print each line with a number (e.g., `1. Inception`).

```python
# Method 1: Read entire file at once
print('=== Method 1: Read All At Once ===')
with open('movies.txt', 'r') as f:
    content = f.read() #f.read: reads the entire file at once
    print(content)

# Method 2: Read line by line with numbers
print('\n=== Method 2: Read Line by Line ===')
with open('movies.txt', 'r') as f:
    for line_number, line in enumerate(f, start=1): # enumerate assigns a number starting at 1
        print(f'{line_number}. {line.strip()}') #removes the \n at the end of each line
```
